```python
# Creating a sample DataFrame
import pandas as pd
data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Neha"],
    "Age": [21, 22, 20, 23, 21],
    "Marks": [85, 90, 78, 92, 88]
}
df = pd.DataFrame(data, index=["S1", "S2", "S3", "S4", "S5"])
print(df)
```

```python
# Selecting a row using loc
print(df.loc["S2"])
```

```python
# Selecting multiple rows using loc
print(df.loc[["S1", "S3", "S5"]])
```

```python
# Selecting a single column using loc
print(df.loc[:, "Marks"])
```

```python
# Selecting multiple columns using loc
print(df.loc[:, ["Name", "Marks"]])
```

```python
# Filtering rows using loc
print(df.loc[df["Marks"] > 85])
```

```python
# Selecting a row using iloc
print(df.iloc[1])
```

```python
# Selecting multiple rows using iloc
print(df.iloc[[0, 2, 4]])
```

```python
# Selecting a single column using iloc
print(df.iloc[:, 2])
```

```python
# Selecting multiple columns using iloc
print(df.iloc[:, [0, 2]])
```

```python
# Selecting specific rows and columns using iloc
print(df.iloc[1:4, 0:2])
```

```python
# Comparing loc and iloc
print("Using loc:")
print(df.loc["S2", "Marks"])
print("Using iloc:")
print(df.iloc[1, 2])
```
